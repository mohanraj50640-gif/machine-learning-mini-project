```python
# stock market movement using ----multinomial navie base
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB
```

```python
x=df['article']
y=df['category']
m=MultinomialNB()
le=LabelEncoder()
```

```python
y_val=le.fit_transform(y)
```

```python
x = df['article']
vector = CountVectorizer()
x_val = vector.fit_transform(x)
```

```python
m.fit(x_val, y_val)
```

```python
while True:
    content=input('enter the article content:')
    test=vector.transform([content])
    predict=m.predict(test)
    print('predicted news:',le.inverse_transform(predict)[0])
```

```python
acc=accuracy_score(y_val,m.predict(x_val))
print('accuracy score:',acc)
```
