```python
#k means
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv('/content/traffic_congestion_pattern_clustering_3000_rows.csv')
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
x=df[['Traffic_Volume_Vehicles_Per_Hour','Average_Speed_kmph','Road_Occupancy_Ratio','Estimated_Travel_Time_Minutes','Congestion_Index']]
```

```python
model=KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(x)
```

```python
#user input
traf=float(input('Traffic_Volume_Vehicles_Per_Hour'))
ave=float(input('Average_Speed_kmph'))
road=float(input('Road_Occupancy_Ratio'))
est=float(input('Estimated_Travel_Time_Minutes'))
cong=float(input('Congestion_Index'))
pred=model.predict([[traf,ave,road,est,cong]])
print('Cluster:',pred[0])
if pred[0]==0:
  print('Low Traffic now')
elif pred[0]==1:
  print('Medium Traffic now')
else:
  print('High Traffic now')
```
