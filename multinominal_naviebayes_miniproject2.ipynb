```python
#socialmedia comment section Analysis ----multinomial navie base
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB
```

```python
df=pd.read_csv('/content/social_media_comment_classification_3000 (1).csv')
df
```

```python
x=df['article']
y=df['category']
m=MultinomialNB()
le=LabelEncoder()
```

```python
y_val=le.fit_transform(y)
```

```python
vector=CountVectorizer()
x_val=vector.fit_transform(x)
```

```python
m.fit(x_val,y_val)
```

```python
while True:
    content=input('Enter Your comments:')
    test=vector.transform([content])
    predict=m.predict(test)
    print('predicted news:',le.inverse_transform(predict)[0])
```

```python
accuracy_score(y_val,m.predict(x_val))
```
