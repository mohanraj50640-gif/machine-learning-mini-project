```python
import pandas as np
import requests
from bs4 import BeautifulSoup
```

```python
url="https://books.toscrape.com"
```

```python
req=requests.get(url)
```

```python
print('status code:',req.status_code)
```

```python
#html to text ----beautifulsoup
soup=BeautifulSoup(req.text,'html.parser')


book=soup.select_one('article.product_pod h3 a')

price=soup.select_one('article.product_pod p.price_color')

print(book.text)
print(price.text)




```

```python
availability=soup.select_one('p.instock.availability')
print(availability.text.strip())
```

```python
####stor as key values like dict
bookdetails={'title':book['title'],
             'price':price.text,
             'availability':availability.text.strip()}
print(bookdetails)
```

```python
print("book details:")
for key,value in bookdetails.items():
  print(f'{key}:{value}')
```

```python
df=np.DataFrame([bookdetails])
df
```

```python
csvconvert=df.to_csv('Mohanrajbookscraping.csv',index=False)
```

```python
xlconvert=df.to_excel("Mohanrajbookscraping.xlsx",index=False)
```
