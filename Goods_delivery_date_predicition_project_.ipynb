```python
#credictcard scam detection with using desicion tree
import pandas  as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score,mean_squared_error
```

```python
df=pd.read_csv('/content/delivery_time_prediction_3000_rows.csv')
df
```

```python
df.head()
```

```python
df.tail()
```

```python
model=DecisionTreeRegressor(max_depth=8,min_samples_leaf=5,random_state=42)
```

```python
le=LabelEncoder()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.values
```

```python
df['a']=le.fit_transform(df['From_City'])
df['b']=le.fit_transform(df['To_City'])
```

```python
y=df['Delivery_Days']
x=df[['a','b','Distance_km']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
r2=r2_score(y_test,y_pred)
print("R2 score:",r2)
mse=mean_squared_error(y_test,y_pred)
print("Mean Squared Error:",mse)
```
